## 1. Comparison operators

Before we can branch, we need to ask questions. Comparison operators compare two values and
answer with a **boolean**: `True` or `False`.

| Operator | Question it asks |
|---|---|
| `==` | are the two values equal? |
| `!=` | are they *not* equal? |
| `>` | is the first greater? |
| `<` | is the first smaller? |
| `>=` | greater than or equal? |
| `<=` | smaller than or equal? |

Note the difference between `=` (assignment) and `==` (comparison) — mixing them up is the
classic beginner error.

````python
print(9 == 9)   # is 9 equal to 9?

a_x = 9
b_x = 8
print(a_x == b_x)   # is a_x equal to b_x?
print(a_x != b_x)   # is a_x NOT equal to b_x?
````

## 2. `if` / `elif` / `else`

### `if`

`if` checks a condition: when it is `True`, the **indented block** below it runs;
otherwise the block is skipped.

````python
if 9 == 9:                     # if the condition holds...
    print("nine equals nine")  # ...this indented line runs
````

**Indentation is the point.** Python uses indentation (one tab / 4 spaces) to know which
lines belong to the `if`. An unindented line is *outside* the condition and always runs:

````python
if 9 == 8:                          # this condition is False...
    print("nine equals eight")      # ...so this indented line never runs
print("this line is NOT inside the if")   # unindented: runs regardless
````

`if` blocks can be nested inside each other — the inner condition is only checked when
the outer one already succeeded:

````python
if 9 == 9:                            # outer condition: True
    if 9 == 8:                        # inner condition: False
        print("nine equals eight")    # never runs
    print("nine equals nine")         # runs: it belongs to the outer if only
````

### `elif` and `else`

- `elif` ("else if") is checked only when the conditions above it failed.
- `else` runs when **nothing** above it matched.

Python reads the chain top to bottom and runs **only the first** branch that matches —
even if a later condition would also be true:

````python
a = 20
if a <= 10:        # False -> skip
    print("first branch")
elif a <= 20:      # True  -> run this one...
    print("second branch")
elif a <= 30:      # ...also True, but never checked: the chain already matched above
    print("third branch")
````

````python
a = 20
if a == 10:
    print("first branch")
elif a == 15:
    print("second branch")
elif a == 30:
    print("third branch")
else:                          # no condition matched...
    print("no branch matched") # ...so else takes over
````

### Separate `if` statements are independent

Unlike an `if`/`elif` chain, two separate `if` statements are **independent cases** —
if both conditions hold, both blocks run:

````python
a = 25
if a <= 10:
    print("group 1: if branch")
elif a <= 20:
    print("group 1: elif branch")
else:
    print("group 1: else branch")

if a == 25:                        # a brand-new, independent check
    print("group 2: if branch")
elif a == 50:
    print("group 2: elif branch")
````

## 3. Logical operators: `and`, `or`, `not`

To combine more than one condition:

- `or` — true when **at least one** condition is true;
- `and` — true only when **all** conditions are true;
- `not` — flips the result.

````python
if 9 == 9 or 9 == 10:    # one true condition is enough for `or`
    print("condition holds")
else:
    print("condition fails")
````

````python
if 9 == 9 and 9 == 10:   # `and` needs ALL conditions to be true
    print("condition holds")
else:
    print("condition fails")
````

````python
if not (9 == 9 and 9 == 10):   # not flips False -> True
    print("condition holds")
else:
    print("condition fails")

if not (9 == 9 or 9 == 10):    # not flips True -> False
    print("condition holds")
else:
    print("condition fails")
````

## 4. `for` loops

When we need to do something many times, we do not copy-paste the same line —
we **loop**. Compare:

````python
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
print("tanakrit")
````

````python
for i in range(10):     # the same thing - shorter, and easy to change to 1000
    print("tanakrit")
````

### `range()` — one, two or three arguments

`range()` generates a sequence of numbers for the loop variable to take:

- `range(stop)` — from 0 up to (but not including) `stop`;
- `range(start, stop)` — choose where to start;
- `range(start, stop, step)` — choose how big each jump is.

````python
for i in range(10):              # i runs 0, 1, ..., 9  (10 iterations)
    print("this is iteration", i)
print("done")                    # not indented -> outside the loop, runs once
````

````python
for i in range(5, 15):       # i runs 5, 6, ..., 14  (still 10 iterations)
    print("i =", i)
````

````python
for i in range(5, 15, 3):    # i runs 5, 8, 11, 14  (steps of 3)
    print("i =", i)
````

A practical use — the 5 times table in two lines:

````python
for i in range(12):
    print("5 *", i + 1, "=", 5 * (i + 1))
````

Loops also combine naturally with `input()` — here, totalling a restaurant bill
(run interactively):

````python
plates = int(input("Number of dishes: "))
total = 0
for x in range(plates):
    dish = input("Dish name: ")
    price = float(input("Price: "))
    total = total + price
    print("Dish", x + 1, ":", price, "baht :", dish)

print("Total price:", total, "baht")
````

### Looping over a list

A `for` loop can walk through **any sequence**, not just `range()`. Looping directly over a
list hands us the *values* one by one — no index bookkeeping needed:

````python
fruit_list = ["apple", "banana", "cherry"]

for fruit in fruit_list:
    print("the list contains:", fruit)
````

The index-based alternative `for i in range(len(list))` also works — useful when you need the
position number itself — but prefer the direct form when you only need the values:

````python
fruit_list_2 = ["apple", "banana", "cherry", "grape", "blueberry", "cherries"]

for i in range(len(fruit_list_2)):   # i is the position: 0, 1, 2, ...
    print(fruit_list_2[i])           # use it to look the value up
````

There is even a one-line form, the **list comprehension** — the expression at the front is
evaluated once per iteration of the loop at the back. We will use it a lot later:

````python
this_list = ["apple", "banana", "cherry"]
[print("i want", x) for x in this_list]
````

### Looping over a dictionary

Looping over a dictionary yields its **keys**; use the key to look up the value,
or use `.items()` to get both at once:

````python
car_dict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
}

for x in car_dict:          # keys
    print(x)

for x in car_dict:          # keys used to fetch values
    print(car_dict[x])

for key, value in car_dict.items():   # both at once
    print(f"{key} : {value}")
````

### Nested loops

A loop inside a loop: for every iteration of the **outer** loop, the **inner** loop runs all
the way through. The classic exercise is drawing star patterns:

````python
for i in range(5):           # 5 rows
    for j in range(5):       # 5 stars per row
        print("*", end="")
    print("")                # row finished -> new line
````

````python
for i in range(5):           # growing triangle:
    for j in range(i + 1):   # row i gets i+1 stars
        print("*", end="")
    print("")
````

````python
for i in range(5):           # shrinking triangle:
    for j in range(5 - i):   # row i gets 5-i stars
        print("*", end="")
    print("")
````

Joining the growing and shrinking triangles gives an arrow:

````python
for i in range(5):
    for j in range(i + 1):
        print("*", end="")
    print("")

for i in range(5):
    for j in range(5 - i):
        print("*", end="")
    print("")
````

Nested loops are not only for drawing — here are all twelve times tables at once:

````python
for i in range(12):
    print("times table for", i + 1)
    for j in range(12):
        print(i + 1, "*", j + 1, "=", (i + 1) * (j + 1))
````

And loops combine with `if` to filter — the even numbers from 1 to 50:

````python
for i in range(1, 51):
    if i % 2 == 0:    # remainder 0 when divided by 2 -> even
        print(i)
````

## 5. `while` loops, `break` and `continue`

A `while` loop keeps repeating **as long as its condition is `True`**. Two keywords control
it from the inside:

- `break` — stop the loop immediately;
- `continue` — skip the rest of this iteration and start the next one.

A guessing game with `while True` + `break` (run interactively — the secret number is 22):

````python
while True:
    print("Guess the number!")
    x = float(input())
    if x == 22:
        print("Correct!")
        break                  # end the loop immediately
    elif x > 22:
        print("Too high - try again")
    else:
        print("Too low - try again")
````

`continue` skips ahead — here it skips the even numbers, leaving only the odd ones:

````python
for i in range(1, 51):
    if i % 2 == 0:
        continue      # even number: skip the print below, go to the next i
    print(i)
````

## 6. Functions

When a piece of logic has fixed steps and gets used repeatedly, we wrap it in a **function**:
define once with `def`, call as many times as we like. A function does nothing until called.

````python
def function_a():           # def + name + () declares a function
    print("hello mama")     # the body is indented, like an if or a loop


function_a()                # ...and this is how we call it
````

Functions accept **parameters** — input values to work with:

````python
def function_b(param_a, param_b):   # two parameters; add more after a comma
    print(param_a + param_b)


function_b(5, 2)
````

### `return` — sending a value back

`print()` only *displays* a result; `return` **hands the value back to the caller**, so it
can be stored, combined and reused in further computation:

````python
def function_c(param_a, param_b):
    return param_a + param_b


print(function_c(2, 10) + 20)              # the returned 12 can be used in arithmetic
print("the result is", function_c(12, 15)) # ...or passed somewhere else
````

A Python function can even return **several values at once** (they come back as a tuple):

````python
def function_d(param_a, param_b):
    a = param_a + param_b
    b = param_a - param_b
    c = param_a * param_b
    d = param_a / param_b
    return a, b, c, d


print(function_d(2, 3))
````

### Controlling parameters

**Default values** make a parameter optional:

````python
def greet(name, message="Hello"):
    return f"{message}, {name}!"


print(greet("Alice"))        # Hello, Alice!  (default used)
print(greet("Bob", "Hi"))    # Hi, Bob!       (default overridden)
````

**Type hints** document what types a function expects and returns. Python does not enforce
them, but they make code far easier to read and let tools catch mistakes:

````python
def greet(name: str) -> str:
    return f"Hello, {name}!"


print(greet("Alice"))
````

### A famous pitfall: mutable default values

A default value is created **once**, when the function is defined — not on every call.
With a mutable default like `lst=[]`, every call that relies on the default shares
**the same list**, and appended values pile up across calls:

````python
def append_to_list(value, lst=[]):
    lst.append(value)
    return lst


print(append_to_list(1))         # [1]
print(append_to_list(2))         # [1, 2]  <- surprise! same list as the first call

print(append_to_list(7, [3, 5])) # [3, 5, 7]  (a list you pass in is fresh)
print(append_to_list(8, []))     # [8]        (so is an explicit empty one)
````

Loops and functions also compose, of course — a times-table function:

````python
def times_table(n):        # which table to print
    for i in range(12):
        print(n, "*", i + 1, "=", n * (i + 1))


times_table(12)
````

## 7. Recursion

A **recursive** function calls itself, breaking a problem into smaller copies of the same
problem until it reaches a **base case** that can be answered directly.

The factorial, $n! = n \times (n-1) \times \dots \times 1$:

````python
def factorial(n):
    if n == 0 or n == 1:            # base case: stop here
        return 1
    else:
        return n * factorial(n - 1) # the function calls itself with a smaller n


factorial(5)
````

The Fibonacci sequence — each number is the sum of the two before it:

````python
def fibonacci(n):
    if n <= 0:
        return "Invalid input"       # guard against bad input
    elif n == 1 or n == 2:           # base cases
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)   # two recursive calls


fibonacci(8)
````

Bonus — a function that combines loops, `print` tricks and string repetition to draw
a tree. Can you work out how it does it?

````python
def tree_making(h: int):
    step = 1
    for i in range(h - 2, -1, -1):
        print(i * " ", "#" * step, sep="")
        step += 2
    print((h - 2) * " ", "#", sep="")


tree_making(6)
````

## Summary

| Topic | Key syntax |
|---|---|
| Comparisons | `== != > < >= <=` → `True`/`False` |
| Branching | `if` / `elif` / `else` + indentation |
| Combining conditions | `and`, `or`, `not` |
| Counted repetition | `for i in range(start, stop, step)` |
| Walking a collection | `for item in list`, `for k, v in d.items()` |
| Open-ended repetition | `while`, with `break` / `continue` |
| Reusable logic | `def f(x, y=default) -> type: return ...` |
| Self-reference | recursion = smaller subproblem + base case |

**Key takeaways**

- Indentation *is* the structure in Python — it decides what belongs to an `if`, a loop or a function.
- An `if`/`elif` chain runs only the **first** matching branch; separate `if`s are independent.
- `return` beats `print` inside functions: a returned value can be reused, a printed one is gone.
- Never use a mutable value (`[]`, `{}`) as a parameter default.

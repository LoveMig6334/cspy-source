# Working with Data 2: Pandas

**Pandas** is one of the most popular Python packages in data science: it makes **data
manipulation** and **data analysis** easy. Where NumPy gives us fast arrays, pandas gives us
**tables** — with named columns, mixed types and a toolbox of analysis methods.

In this lecture:

1. **Pandas data structures** — `Series` (one column) and `DataFrame` (a whole table).
2. **Reading data** — loading a real dataset (the Titanic passenger list) from a CSV file.
3. **Selecting data** — by position (`iloc`/`iat`), by label (`loc`/`at`), and by condition
   (boolean indexing).
4. **Aggregation** — descriptive statistics, counting categories, and `groupby`.
5. **Descriptive statistics on a second dataset** — fuel-economy data, correlations,
   top-N queries.

```python
import pandas as pd   # the universal convention, like np for numpy
```

## 1. Pandas data structures

### Series — a single column

A `Series` stores data as a 1-D array — think of **one column** in a spreadsheet.
Note the index numbers pandas attaches on the left:

```python
ser = pd.Series([5, 10, 15])

print("Data dimension:", ser.ndim)   # .ndim - number of dimensions
print("Data shape:", ser.shape)      # .shape - structure, like a NumPy array
ser
```

### DataFrame — a whole table

A `DataFrame` is several Series side by side — a table with rows and **named columns**,
looking exactly like a spreadsheet. One way to build one: a matrix of values plus a list of
column names:

```python
df_columns = ["firstname", "surname", "age"]   # the column names

df_values = [                                  # the table body is just a matrix
    ["tanakrit", "saiphan", 17],
    ["nayos", "suwatano", 18],
    ["pannathorn", "krudthong", 18],
]

df = pd.DataFrame(columns=df_columns, data=df_values)
df
```

Equivalently, build it from a dictionary — keys become column names, values become the
columns:

```python
data = {
    "firstname": ["tanakrit", "mikey", "pannathorn"],
    "surname": ["saiphan", "suwatano", "krudthong"],
    "age": [17, 18, 18],
}

df = pd.DataFrame(data)
df   # same table as above
```

## 2. Reading data from a file

In practice we rarely type DataFrames by hand — we **read** them from files. Pandas can read
CSV, JSON, Excel and many more formats
([full list here](https://pandas.pydata.org/pandas-docs/stable/user_guide/io.html)).

We load the classic **Titanic** dataset — one row per passenger:

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
)
df
```

891 rows is too much to read at once. The first-look toolkit:

- `head(n)` — the first n rows (default 5);
- `tail(n)` — the last n rows;
- `sample(n)` — n random rows;
- `info()` — column names, types, and **how many values are missing**.

```python
df.head(1)   # the very first row
```

```python
df.tail(2)   # the last two rows
```

```python
df.sample(2)   # two random rows
```

```python
df.info()   # column types and missing-value counts (look at Age and Cabin!)
```

## 3. Selecting data by position or label

### Getting — plain indexing

A Series indexes like a list; a DataFrame sliced with a range of numbers returns those rows:

```python
print(ser[1])    # one element of the Series
ser[0:2]         # a slice of the Series
```

```python
df[56:60]   # rows 56-59 of the DataFrame
```

### Selecting by position: `iloc` and `iat`

`iloc[row, column]` selects by **integer position** — row and column numbers, both starting
at 0. `iat` is its faster single-value cousin (it requires *both* a row and a column, while
`iloc` can take just a row):

```python
df.head()
```

```python
print(df.iloc[0, 3])   # row 0, column 3
print(df.iloc[2, 8])   # row 2, column 8
print(df.iat[0, 3])    # same as the first lookup, single-value access
```

### Selecting by label: `loc` and `at`

`loc[row_label, column_name]` selects by **label** — here the row labels are the index
numbers and the column labels are the column names:

```python
print(df.loc[0, "Name"])   # row 0, column "Name"
print(df.at[1, "Name"])    # row 1, column "Name" - single-value access
```

### Selecting subsets

Both `iloc` and `loc` extend naturally from single values to whole blocks of the table:

```python
df.iloc[0]   # all of row 0 - one value per column
```

```python
df.iloc[:, 5]   # every row, but only column 5
```

```python
df.iloc[0:4]   # rows 0-3
```

```python
df.iloc[0:4, 0:4]   # rows 0-3, columns 0-3
```

```python
df.iloc[[0, 4], [0, 4]]   # exactly rows 0 and 4, columns 0 and 4
```

```python
df.loc[0:4]   # by label: rows 0-4 (note: loc INCLUDES the endpoint!)
```

```python
df.loc[0:2, "PassengerId":"Age"]   # rows 0-2, columns PassengerId through Age
```

```python
df.loc[0:4, ["PassengerId", "Age"]]   # rows 0-4, just these two columns
```

### Boolean indexing

The boolean-mask trick from NumPy works on tables too. A comparison applied to a Series or a
column produces `True`/`False` per row; indexing with that mask keeps only the `True` rows:

```python
ser == 5   # which elements equal 5?
```

```python
print(ser[ser == 5])    # keep only the matching elements
print(ser[ser < 15])    # all elements below 15
```

The same idea finds all female passengers on the Titanic:

```python
df["Sex"] == "male"   # one True/False per row
```

```python
df[df["Sex"] == "female"]   # keep only the rows where the mask is True
```

## 4. Aggregation

**Aggregating** summarises many rows into a few numbers — means, maxima, minima, most frequent
values, and so on. First we narrow the table to the columns of interest:

```python
df2 = df[["PassengerId", "Sex", "Age", "Ticket", "Fare"]]   # whole table, chosen columns
df2.head()
```

```python
df2.dtypes   # what type is each column?
```

### Numerical data

For descriptive statistics we can call individual functions — or `describe()` for everything
at once: mean, standard deviation, min, max and percentiles.

```python
print(f"Youngest passenger: {df2['Age'].min()}")
print(f"Oldest passenger:   {df2['Age'].max()}")
print(f"Average age:        {df2['Age'].mean()}")
print(f"Median age:         {df2['Age'].median()}")
```

```python
df2.describe()   # all of the above (and more) in one call
```

### Categorical data

For category-like columns we ask different questions: what distinct values exist (`unique`),
how many distinct values (`nunique`), and how often each occurs (`value_counts`):

```python
df["Ticket"].unique()   # the distinct ticket codes (no repeats)
```

```python
df["Ticket"].nunique()   # how many distinct values
```

```python
df["Ticket"].value_counts()   # how often each value occurs
```

### Categorical × numerical: `groupby`

`groupby()` splits the table into groups by category, applies a statistic per group, and
combines the results
([split-apply-combine](https://pandas.pydata.org/docs/user_guide/groupby.html)).
The oldest, youngest and average passenger — separately for each sex:

```python
df2.groupby(["Sex"])["Age"].max()    # oldest passenger per group
```

```python
df2.groupby(["Sex"])["Age"].min()    # youngest per group
```

```python
df2.groupby(["Sex"])["Age"].mean()   # average age per group
```

```python
df2.groupby(["Sex"])["Fare"].mean()  # average ticket fare per group
```

## 5. Descriptive statistics on a second dataset

Let's put the whole toolkit to work on a fresh dataset: fuel-economy figures for 234 cars
(`mpg`). Same first-look routine as before:

```python
mpg = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/tutorial/master/mpg.csv")
mpg.head()
```

```python
print(mpg.shape)   # (rows, columns)
mpg.info()
```

The table mixes text and number columns, so for table-wide statistics we pass
`numeric_only=True` — averaging a manufacturer name makes no sense:

```python
mpg.mean(numeric_only=True)     # column averages
```

```python
mpg.median(numeric_only=True)   # column medians
```

Counting categories: how many models per manufacturer, and per vehicle class? Note that
`mpg.class` would be a **syntax error** — `class` is a Python keyword — so bracket notation
`mpg["class"]` is required for this column:

```python
mpg.manufacturer.value_counts()   # dot notation works for ordinary column names
```

```python
mpg["class"].value_counts()   # bracket notation - "class" clashes with the keyword
```

`describe()` again — and a trick: transposing it with `.T` puts one **row per column**,
which is much easier to scan when there are many columns:

```python
mpg.describe()
```

```python
mpg.describe().T   # transposed: one row per column
```

### Correlation

`corr()` measures how strongly each pair of numeric columns moves together:
**+1** = perfectly proportional, **−1** = perfectly inversely proportional, **0** = no
relationship. City and highway fuel economy, unsurprisingly, are strongly correlated:

```python
mpg.corr(numeric_only=True)
```

### Top-N queries

`nlargest` / `nsmallest` sort by a column and keep the extremes — the 5 most fuel-efficient
cars in the city, and the 5 thirstiest on the highway:

```python
mpg.nlargest(5, columns="cty")    # top 5 by city fuel economy
```

```python
mpg.nsmallest(5, columns="hwy")   # bottom 5 by highway fuel economy
```

## Summary

| Task | Function / syntax |
|---|---|
| One column / whole table | `pd.Series([...])` / `pd.DataFrame(dict)` |
| Load a file | `pd.read_csv(path_or_url)` |
| First look | `head`, `tail`, `sample`, `info`, `shape`, `dtypes` |
| Select by position | `df.iloc[rows, cols]`, `df.iat[r, c]` |
| Select by label | `df.loc[rows, "col_a":"col_b"]`, `df.at[r, "col"]` |
| Filter by condition | `df[df["col"] == value]` |
| Summary statistics | `mean`, `median`, `min`, `max`, `describe()` |
| Categories | `unique`, `nunique`, `value_counts` |
| Per-group statistics | `df.groupby("cat")["num"].mean()` |
| Relationships / extremes | `corr(numeric_only=True)`, `nlargest`, `nsmallest` |

**Key takeaways**

- A DataFrame is a spreadsheet you can program: Series = column, DataFrame = table.
- Always start with `head()` + `info()` — know your columns, types and missing values
  before computing anything.
- `iloc` selects by *position*, `loc` by *label* — and `loc` slices **include** the endpoint.
- Boolean masks carry over from NumPy and are the main way to filter rows.
- `groupby` answers "statistic X per category Y" in a single line.

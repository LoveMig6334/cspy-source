## 1. The `print()` statement

`print()` displays values on the screen: text, numbers, variables, or the result of a function.
It is the first tool you reach for to *see* what your program is doing.

```python
print("Any text you like")
```

You can print several values at once by separating them with commas — `print()` inserts a
space between them automatically. The separator can be changed with the `sep=` argument:

```python
print("text 1", "text 2")          # separated by a space (the default)
print("text 1", "text 2", sep="")  # no separator at all
```

`print()` is not limited to text — it can display variables too:

```python
a = 15
b = 16
print(a, b)
print(a)
```

## 2. f-strings: variables inside text

An **f-string** (`f"..."`) lets us embed the value of a variable directly inside a string
using curly braces `{variable}`. This is the cleanest way to combine text and values:

```python
age = 16
name = "Ram"
print(f"My name is {name} and I am {age} years old. {age} is my favorite number.")
```

## 3. Working with strings

Python strings come with **built-in methods** for cleaning and reshaping text.
The most common ones:

| Method | What it does |
|---|---|
| `upper()` / `lower()` | convert to UPPER / lower case |
| `strip()` / `lstrip()` / `rstrip()` | remove whitespace (both sides / left / right) |
| `replace(old, new)` | replace a substring |
| `split(separator)` | cut the string into a list of pieces |

```python
text = "hello world"
print(text.upper())   # all characters to upper case -> "HELLO WORLD"

text = "HELLO WORLD"
print(text.lower())   # all characters to lower case -> "hello world"
```

```python
text = "   Hello World   "
print(text.strip())   # "Hello World"     - whitespace removed on both sides
print(text.lstrip())  # "Hello World   "  - removed on the left only
print(text.rstrip())  # "   Hello World"  - removed on the right only
```

```python
text = "I like apples"
result = text.replace("apples", "bananas")
print(result)  # "I like bananas"
```

```python
text = "apple,banana,cherry"
print(text.split(","))   # ['apple', 'banana', 'cherry']

text2 = "Hello World"
print(text2.split())     # with no argument it splits on whitespace -> ['Hello', 'World']
```

## 4. Line endings

By default every `print()` ends with a newline, so consecutive calls appear on separate lines.
Three ways to control this:

- `end=""` — stay on the same line after printing;
- `\n` inside a string — insert a line break manually;
- `end="\r"` — return to the *start of the same line*, so the next print overwrites it
  (the classic trick behind progress indicators).

```python
print("Hello")
print("Python")   # each print() starts a new line by default
```

```python
print("Hello", end="")
print(" all of the students")   # continues on the same line
```

```python
print("hello\nstudent")   # \n breaks the line in the middle of a string
```

```python
print("Hello", end="\r")   # \r returns to the start of the line...
print("Python", end="\r")  # ...so this overwrites "Hello"...
print("Hello me")          # ...and this overwrites "Python"
```

Combining `\r` with a loop gives us a simple progress display that updates in place:

```python
import time

steps = range(1, 200)
for i in steps:
    progress = (i / len(steps)) * 100
    print(f"{progress:.2f} %", end="\r")   # overwrite the same line each iteration
    time.sleep(0.02)
```

## 5. Variables and data types

A **variable** is a container for a value. We store a value in a variable with the `=` sign,
then refer to it by name whenever we need that value.

```python
number = 10                       # declare a variable
print(number)                     # a variable can be printed
print("The number is", number)   # and combined with text
```

We can create as many variables as we like, and do arithmetic with them:

```python
first = 35
second = 70
print(first, second)   # print more than one variable by separating with commas

added = first + 99
print(added)
print(first)           # note: `first` itself is unchanged
```

### Python's data types

Every value in Python has a **type**. The full catalogue:

* Text: `str`
* Numbers: `int`, `float`, `complex`
* Sequences: `list`, `tuple`, `range`
* Mapping: `dict`
* Sets: `set`, `frozenset`
* Boolean: `bool`
* Binary: `bytes`, `bytearray`, `memoryview`
* No value: `NoneType`

We will meet the sequence and mapping types properly in week 4. For now, the basic four:

```python
x_integer = 123                        # int: a whole number
x_float = 456.789                      # float: a number with a decimal point
x_string = "Good morning every one"    # str: text, always wrapped in ' or "
x_boolean = True                       # bool: logical value, True or False

print(x_integer, x_float, x_string, x_boolean)
```

To check the type of any value, use the `type()` function:

```python
print(type(x_integer))
print(type(x_float))
print(type(x_string))
print(type(x_boolean))
```

## 6. Getting user input

The `input()` function pauses the program and waits for the user to type something.
Whatever they type comes back as a **string**, which we usually store in a variable.
We can also pass a prompt message so the user knows what to enter.

> These cells wait for keyboard input — run them interactively.

```python
name = input("Enter your name: ")   # the prompt explains what to type
print("Hi", name)
```

`input()` always returns a string. To treat the answer as a number we must convert it,
e.g. with `int()`:

```python
input_number = int(input("Enter a number: "))   # convert the typed text to an int
print("Your number is", input_number)
```

## 7. Converting between types

Sometimes data arrives as one type and we need another. The conversion functions are simply
the type names used as functions: `int(value)`, `float(value)`, `str(value)`.

```python
integer = 35
print(type(integer))

new_number = float(integer)   # convert int -> float
print(type(new_number))
print(new_number)
```

**Careful:** converting something that cannot be converted raises an error.
`float("abc")` has no sensible answer, so Python stops with a `ValueError` —
run the next cell and read the error message:

```python
var = float("abc")   # ValueError: could not convert string to float
print(var)
```

## 8. Operators

Python supports all the usual arithmetic operations:

| Operator | Meaning |
|---|---|
| `+` | addition |
| `-` | subtraction |
| `*` | multiplication |
| `/` | division |
| `**` | exponentiation |
| `//` | floor division (divide, drop the remainder) |
| `%` | modulo (the remainder of a division) |

```python
a = 10
b = 20
print(f"a = {a}, b = {b}")
print("sum:", a + b, " difference:", a - b, " product:", a * b, " quotient:", a / b)
```

A small worked example — the area of a triangle, $A = \frac{1}{2} \times b \times h$:

```python
base = 10
height = 15

area = (1 / 2) * base * height

print("base =", base)
print("height =", height)
print("area =", area)
```

And a rectangle, $A = w \times h$:

```python
w = 10
h = 20
area = w * h

print("width =", w)
print("height =", h)
print("area =", area)
```

### Assignment operators

Assignment operators combine an arithmetic operation with assignment in one step.
`x += 3` is shorthand for `x = x + 3`, and the same pattern works for every operator:

| Operator | Equivalent to |
|---|---|
| `x += y` | `x = x + y` |
| `x -= y` | `x = x - y` |
| `x *= y` | `x = x * y` |
| `x /= y` | `x = x / y` |
| `x %= y` | `x = x % y` |
| `x //= y` | `x = x // y` |
| `x **= y` | `x = x ** y` |

```python
x = 5
x += 3    # x = x + 3
print(x)  # 8

x = 5
x -= 2    # x = x - 2
print(x)  # 3

x = 5
x *= 2    # x = x * 2
print(x)  # 10

x = 10
x /= 2    # x = x / 2
print(x)  # 5.0  (division always gives a float)
```

```python
x = 10
x %= 3    # remainder of 10 / 3
print(x)  # 1

x = 10
x //= 3   # floor division
print(x)  # 3

x = 2
x **= 3   # 2 to the power 3
print(x)  # 8
```

### Operators on strings

`+` and `*` also work on strings — `+` **concatenates** two strings, and `*` repeats a
string a whole number of times:

```python
first_name = "Tanakrit"
last_name = "Saiphan"

print(first_name + last_name)   # + joins strings together
print(3 * first_name)           # * repeats a string
```
